### Задание 1

Посчитайте частные производные функций:

1) $f(x,y)=2x^2y^3 + 1/x + y^2x + 7$

2) $f(x,y)=x^2y - sin(xy) + cos(x^2) + 6y$

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
x,y = sp.symbols('x y',real=True)
f1 = 2*x**2*y**3 + 1/x + y**2*x + 7
f2 = x**2*y-sp.sin(x*y)+sp.cos(x**2)+6*y
for expr in [f1,f2]:
    display(sp.diff(expr,x),sp.diff(expr,y))
```

## Уровень 1:

### Задание 2

Вычислите градиент функции:

$f(x,y)=2x^2y^3 + 1/x + y^2x + 7$

в точке $(1,2)$

```python
gradient = sp.Matrix([sp.diff(f1,x),sp.diff(f1,y)])
value = gradient.subs({x:1,y:2})
assert value == sp.Matrix([35,28])
display(gradient,value)
```

### Задание 3

Постройте трехмерный график функции

$f(x)=x_2^2-x_1^2 x_2 + 2 x_1 x_2$

```python
grid = np.linspace(-3,3,120)
X1,X2 = np.meshgrid(grid,grid)
Z = X2**2-X1**2*X2+2*X1*X2
fig = plt.figure(figsize=(10,7))
ax = fig.add_subplot(111,projection='3d')
surface = ax.plot_surface(X1,X2,Z,cmap='viridis')
ax.set(xlabel='x₁',ylabel='x₂',zlabel='f',title='f = x₂² − x₁²x₂ + 2x₁x₂')
fig.colorbar(surface,ax=ax,shrink=0.6); plt.show()
```
