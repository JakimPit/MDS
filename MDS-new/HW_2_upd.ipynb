### Задание 1

 - Создайте матрицу 3 на 3, заполненную числами от 1 до 9
 - Прибавьте к ней матрицу с единицами на диагонале
 - Умножьте полученную матрицу на вектор (1, 2, 3)
 - Вычислить транспонированную матрицу $A^T$ от полученной в результате прибавления матрицы
 - Посчитайте обратную матрицу к полученной в результате прибавления матрице


````python
import numpy as np
A = np.arange(1, 10).reshape(3,3)
B = A + np.eye(3)
product = B @ np.array([1,2,3])
transpose = B.T
inverse = np.linalg.inv(B)
assert np.allclose(B @ inverse, np.eye(3))
assert np.array_equal(product, [15,34,53])
for title, value in [('A',A), ('A + I',B), ('Произведение',product), ('Транспонирование',transpose), ('Обратная',inverse)]:
    print(title, '\n', value)
````

### Задание 2

Найдите собственные значения и собственные вектора матрицы:
```
m = np.array([[2, 2],
              [1, 3]])
print(m)
```



````python
m = np.array([[2,2],[1,3]])
values, vectors = np.linalg.eig(m)
assert np.allclose(m @ vectors, vectors @ np.diag(values))
assert np.allclose(np.sort(values), [1,4])
print('Собственные значения:', values)
print('Собственные векторы — столбцы:', vectors)
````

### Задание 3


Придумайте матрицу, у которой определитель = 1, и все элементы не равны 0.

P.S. Матрица должна быть ненулевой и размерностью не менее 2! 

````python
C = np.array([[1,1],[1,2]])
assert np.all(C != 0) and np.isclose(np.linalg.det(C), 1)
print(C, 'det =', np.linalg.det(C))
````
